```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
htmap = pd.read_csv('/hdd/exvivo_atac/scenic+/first_cluster_PCC.csv')
htmap
```

```python
# remove the "symbol" from the dataframe

htmap_rn = htmap.iloc[:,0]
htmap_rn.name = None
htmap_rn
```

```python
# assign the gene symbol as the rownames
htmap = htmap.iloc[:,1:]
htmap.index = htmap_rn
htmap
```

```python
# heatmap visualization
sns.heatmap(htmap)
```

```python
# using mask
mask = np.zeros_like(htmap)
mask[np.triu_indices_from(mask)] = True
with sns.axes_style("white"):
    ax = sns.heatmap(htmap, mask=mask, vmax=1, square=True, vmin = -1, center = 0, cmap = 'vlag')

```
