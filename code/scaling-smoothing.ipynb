```python
from sklearn.neighbors import KDTree
X = scaled.obsm["X_pca"]
print(X)
```

```python
K = 31  # number of nearest neighbors + 1
tree = KDTree(X, leaf_size=40, metric='euclidean')
# For finding K neighbors of P1 with shape (1, 3)
# distances, indices = tree.query(X[:1], K)  # slice the data by [a:b] to retain the array's dimension, because it only allows 2D input
indices = tree.query(X, K, return_distance=False)
print(indices)
# print(distances)
```

let's save the knn indices information into a .npy file

```python
np.save("/hdd/edward/the-project/results/pipeline/knn_indices.npy", indices)
```

let's substitute unsmoothed data with smoothed ones

```python
for i in range(len(scaled.X)):
    smoothed.X[i] = np.mean([scaled.X[j] for j in indices[i]], axis=0)
```
