```python
import scanpy as sc
import scanpy.external as sce
import numpy as np
import gc
```

```python
adata = sc.read_h5ad("discrete_cancer_stages_annotated_magic.h5ad")
adata
```

```python
sc.set_figure_params(fontsize=18, figsize="6,6")
sc.pp.highly_variable_genes(adata, layer='magic_log', n_top_genes=2500)
sc.pl.highly_variable_genes(adata, save="_NE_magic.svg")
adata
```

```python
adata.X
```

```python
adata.layers['magic_scale'] = adata.layers['magic_log']
adata
```

```python
adata.layers["magic_log"]
```

```python
sc.pp.scale(adata, layer='magic_scale', max_value=10)
adata
```

```python
adata.layers["magic_scale"]
```

```python
sc.tl.pca(adata, svd_solver='arpack', use_highly_variable=True, layer='magic_scale')
adata
```

```python
sc.pl.pca(adata, color=['MYCN', 'PHOX2B'], save="_NE_markers_magic.svg")
```

```python
sc.pl.pca_variance_ratio(adata, n_pcs=50, log=True, save="_NE_magic.svg")
```

```python
# use_rep='X_pca_harmony' denotes I'm using corrected PCA to compute neighbors
sc.pp.neighbors(adata, n_neighbors=15, n_pcs=50)
adata
```

```python
sc.tl.leiden(adata)
adata
```

```python
# paga requires leiden or louvain to run first, because it is a TI algorithm which uses clusters to infer trajectories
sc.tl.paga(adata)
sc.set_figure_params(fontsize=20)
sc.pl.paga(adata, frameon=False, title="PAGA", fontsize=15, save='_magic.svg')  # remove `plot=False` if you want to see the coarse-grained graph
adata
```

```python
# sc.tl.umap(adata)
# use the paga embedding
sc.tl.umap(adata, init_pos='paga')
adata
```

```python
sc.set_figure_params(fontsize=20, figsize=None)
sc.pl.umap(adata, color=['leiden'], legend_fontsize=10, save='_leiden_magic.svg')
```
