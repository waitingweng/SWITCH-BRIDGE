add patient ID annotation

```python
id_annotation_list = []
for barcode in adata.obs_names:
    id = barcode.split("_")[0]
    id_annotation_list.append(id)
```

```python
adata.obs["sample_ID"] = id_annotation_list
adata.obs[["sample_ID"]]
```

```python
male_ids = ["T10", "T27", "T40", "T69", "T71", "T162", "T175", "T188", "T230"]
female_ids = ["T19", "T34", "T44", "T75", "T92", "T200", "T214"]
```

```python
gender_annotation_list = []
# male = ["T10", "T27", "T40", "T69", "T71", "T162", "T174", "T188", "T230"]  # we do not include T10 in our dataset
for i in adata.obs["sample_ID"]:
    if i[0] == 'F':
        gender_annotation_list.append("NA")
    elif i in male_ids:
        gender_annotation_list.append("M")
    else:
        gender_annotation_list.append("F")

print(gender_annotation_list[:10])
```

```python
adata.obs["Gender"] = gender_annotation_list
adata.obs["Gender"]
```

```python
stage1 = ["T44", "T175", "T188"]
stage3 = ["T19", "T200", "T214", "T230"]
stage4 = ["T10", "T27", "T34", "T40", "T69", "T75", "T92", "T162"]
stage4S = ["T71"]
```

```python
INSS_annotation_list = []
for i in adata.obs["sample_ID"]:
    if i in stage1:
        INSS_annotation_list.append("I")
    elif i in stage3:
        INSS_annotation_list.append("III")
    elif i in stage4:
        INSS_annotation_list.append("IV")
    elif i in stage4S:
        INSS_annotation_list.append("4S")
    else:
        INSS_annotation_list.append("NA")
        
print(INSS_annotation_list[:10])
```

```python
adata.obs["INSS_stage"] = INSS_annotation_list
adata.obs["INSS_stage"]
```
