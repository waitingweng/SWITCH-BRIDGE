The simplest and fastest method to do so is the t-test.

for INSS stage

```python
sc.tl.rank_genes_groups(adata, groupby="INSS_stage", method='t-test')
sc.pl.rank_genes_groups(adata, n_genes=50, sharey=False)
```

for discrete_cancer_stage

```python
sc.tl.rank_genes_groups(adata, groupby="discrete_cancer_stage", method='t-test')
sc.pl.rank_genes_groups(adata, n_genes=50, sharey=False)
```

Wilcoxon rank-sum (Mann-Whitney-U) test, recommended

```python
sc.tl.rank_genes_groups(adata, "INSS_stage", method='wilcoxon')
sc.pl.rank_genes_groups(adata, n_genes=50, sharey=False)
```

```python
sc.tl.rank_genes_groups(adata, "discrete_cancer_stage", method='wilcoxon')
sc.pl.rank_genes_groups(adata, n_genes=50, sharey=False)
```

rank genes using logistic regression.
The essential difference is that here, we use a multi-variate appraoch whereas conventional differential tests are uni-variate.

```python
sc.tl.rank_genes_groups(adata, 'INSS_stage', method='logreg')
sc.pl.rank_genes_groups(adata, n_genes=50, sharey=False)
```

```python
sc.tl.rank_genes_groups(adata, 'discrete_cancer_stage', method='logreg')
sc.pl.rank_genes_groups(adata, n_genes=50, sharey=False)
```
