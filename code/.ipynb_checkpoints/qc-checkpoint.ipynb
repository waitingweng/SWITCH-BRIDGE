```python
import scanpy as sc
import skimage
```

```python
# verbosity: errors (0), warnings (1), info (2), hints (3)
sc.settings.verbosity = 3
sc.logging.print_header()
sc.settings.set_figure_params(dpi=80, facecolor='white')  # dpi is the resolution of rendered figures
```

take a look at the anndata

```python
# large_root = r'/Users/edwerd/Desktop/single-cell-RNA-sequencing-tutorial-1/science-advance-paper-practice/nb_GOSH_cellxgene.h5ad'
adata = sc.read_h5ad("annotated.h5ad")
print(adata.obs_names)
print(adata.var_names)
adata
```

start pre-processing

```python
sc.set_figure_params(fontsize=18, figsize=None)
```

```python
# Show those genes that yield the highest fraction of counts in each single cell, across all cells
sc.pl.highest_expr_genes(adata, n_top=20, save='.svg')  # plot highest expressing genes
adata
```

```python
sc.pp.filter_cells(adata, min_genes=200)  # filtering
sc.pp.filter_genes(adata, min_cells=3)
```

```python
# annotate the group of mitochondrial genes as 'mt' for QC
adata.var['mt'] = adata.var_names.str.startswith('MT-')
sc.pp.calculate_qc_metrics(adata, qc_vars=['mt'], percent_top=None, log1p=False, inplace=True)
adata
```

```python
sc.pl.violin(
    adata,
    ["n_genes_by_counts", "total_counts", "pct_counts_mt"],
    jitter=0.4,
    multi_panel=True,
    save='.svg'
)
```

```python
sc.pl.scatter(adata, x='total_counts', y='pct_counts_mt', save='_pct_counts_mt_vs_total_counts.svg')
sc.pl.scatter(adata, x='total_counts', y='n_genes_by_counts', save='_n_genes_by_counts_vs_total_counts.svg')
```

```python
sc.pl.scatter(adata, "total_counts", "n_genes_by_counts", color="pct_counts_mt", save='_n_genes_by_counts_vs_total_counts_colored.svg')
```

```python
adata = adata[adata.obs.n_genes_by_counts < 7000, :]
# according to https://www.biorxiv.org/content/10.1101/2020.02.20.958793v1
# human's mitochondrial ratio has to be higher than 5% (the standard for mice)
# For human tissues, the 5% threshold should be reconsidered as it fails to accurately discriminate between healthy and low-quality cells in 29.5% (13 of 44) tissues analyzed.
adata = adata[adata.obs.pct_counts_mt < 10, :]
adata
```

```python
# remove doublet cells
sc.pp.scrublet(adata, batch_key="sample_ID", n_prin_comps=50)
```

saving the raw data

```python
adata = adata[adata.obs.predicted_doublet == False]
adata
```

```python
# saving the raw data to a file
adata.write("raw.h5ad")
```
