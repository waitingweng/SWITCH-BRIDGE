```python
sdata = sc.read_h5ad("cell_type_classified.h5ad")
sdata
```

add gender

```python
male_ids = ["T10", "T27", "T40", "T69", "T71", "T162", "T175", "T188", "T230"]
female_ids = ["T19", "T34", "T44", "T75", "T92", "T200", "T214"]
```

```python
gender_annotation_list = []
# male = ["T10", "T27", "T40", "T69", "T71", "T162", "T174", "T188", "T230"]  # we do not include T10 in our dataset
for i in sdata.obs["sample_ID"]:
    if i[0] == 'F':
        gender_annotation_list.append("NA")
    elif i in male_ids:
        gender_annotation_list.append("M")
    else:
        gender_annotation_list.append("F")

print(gender_annotation_list[:10])
```

```python
sdata.obs["Gender"] = gender_annotation_list
sdata.obs["Gender"]
```

add age

```python
age_dict = {"T10": "1185", "T19": "790", "T27": "1215", 
            "T34": "1155", "T40": "86", "T44": "66", 
            "T69": "880", "T71": "5", "T75": "280", 
            "T92": "1245", "T162": "1000", "T175": "23", 
            "T188": "113", "T200": "910", "T214": "118", 
            "T230": "1700"}
```

```python
age_annotation_list = []
for i in sdata.obs["sample_ID"]:
    if i in age_dict:
        age_annotation_list.append(age_dict[i])
    else:
        age_annotation_list.append("NA")
        
print(age_annotation_list[:10])
```

```python
sdata.obs["Age"] = age_annotation_list
sdata.obs["Age"]
```

add Shimada favorable/unfavorable

```python
favorable = ["T19", "T27", "T40", "T44", "T71", "T75", "T188"]
unfavorable = ["T10", "T34", "T69", "T92", "T162", "T175", "T200", "T214", "T230"]
```

```python
Shimada_annotation_list = []
for i in sdata.obs["sample_ID"]:
    if i in favorable:
        Shimada_annotation_list.append("FH")
    elif i in unfavorable:
        Shimada_annotation_list.append("UH")
    else:
        Shimada_annotation_list.append("NA")
        
print(Shimada_annotation_list[:10])
```

```python
sdata.obs["Shimada"] = Shimada_annotation_list
sdata.obs["Shimada"]
```

add INSS stages

```python
stage1 = ["T44", "T175", "T188"]
stage3 = ["T19", "T200", "T214", "T230"]
stage4 = ["T10", "T27", "T34", "T40", "T69", "T75", "T92", "T162"]
stage4S = ["T71"]
```

```python
INSS_annotation_list = []
for i in sdata.obs["sample_ID"]:
    if i in stage1:
        INSS_annotation_list.append("I")
    elif i in stage3:
        INSS_annotation_list.append("III")
    elif i in stage4:
        INSS_annotation_list.append("IV")
    elif i in stage4S:
        INSS_annotation_list.append("4S")
    else:
        INSS_annotation_list.append("NA")
        
print(INSS_annotation_list[:10])
```

```python
sdata.obs["INSS_stage"] = INSS_annotation_list
sdata.obs["INSS_stage"]
```

add COG risk group

```python
high = ["T10", "T27", "T34", "T69", "T71", "T75", "T92", "T162", "T200", "T214", "t230"]
intermediate = ["T19", "T40"]
low = ["T44", "T175", "T188"]
```

```python
COG_annotation_list = []
for i in sdata.obs["sample_ID"]:
    if i in high:
        COG_annotation_list.append("high-risk")
    elif i in intermediate:
        COG_annotation_list.append("intermediate-risk")
    elif i in low:
        COG_annotation_list.append("low-risk")
    else:
        COG_annotation_list.append("NA")
        
print(COG_annotation_list[:10])
```

```python
sdata.obs["COG_risk_group"] = COG_annotation_list
sdata.obs["COG_risk_group"]
```

add distant metastasis

```python
bone_and_marrow = ["T10", "T92", "T162"]
no = ["T19", "T44", "T162", "T175", "T188", "T200", "T214", "T230"]
axillary_lymph_nodes = ["T27", "T40"]
bone = ["T34", "T75"]
bone_and_craniocerebral = ["T69"]
liver = ["T71"]
```

```python
metastasis_annotation_list = []
for i in sdata.obs["sample_ID"]:
    if i in bone_and_marrow:
        metastasis_annotation_list.append("Bone_and_bone_marrow")
    elif i in no:
        metastasis_annotation_list.append("no")
    elif i in axillary_lymph_nodes:
        metastasis_annotation_list.append("Axillary_lymph_nodes")
    elif i in bone:
        metastasis_annotation_list.append("Bone")
    elif i in bone_and_craniocerebral:
        metastasis_annotation_list.append("Bone_and_craniocerebral")
    elif i in liver:
        metastasis_annotation_list.append("Liver")
    else:
        metastasis_annotation_list.append("NA")
        
print(metastasis_annotation_list[:10])
```

```python
sdata.obs["Distant_metastasis"] = metastasis_annotation_list
sdata.obs["Distant_metastasis"]
```

add MYCN-amp

```python
amplified = ["T162", "T200", "T230"]
not_amplified = ["T10", "T19", "T27", "T34", "T40", "T44", 
                 "T69", "T71", "T75", "T92", "T175", "T188", "T214"]
```

```python
MYCN_annotation_list = []
for i in sdata.obs["sample_ID"]:
    if i in amplified:
        MYCN_annotation_list.append("amplified")
    elif i in not_amplified:
        MYCN_annotation_list.append("not_amplified")
    else:
        MYCN_annotation_list.append("NA")
        
print(MYCN_annotation_list[:10])
```

```python
sdata.obs["MYCN-amp"] = MYCN_annotation_list
sdata.obs["MYCN-amp"]
```

add survival/death

```python
survival = ["T10", "T19", "T34", "T40", "T44", 
            "T69", "T75", "T92", "T162", "T175",
            "T188", "T200", "T214", "T230"]
death = ["T27", "T71"]
```

```python
survival_annotation_list = []
for i in sdata.obs["sample_ID"]:
    if i in survival:
        survival_annotation_list.append("survival")
    elif i in death:
        survival_annotation_list.append("death")
    else:
        survival_annotation_list.append("NA")
        
print(survival_annotation_list[:10])
```

```python
sdata.obs["Survival/Death"] = survival_annotation_list
sdata.obs["Survival/Death"]
```
