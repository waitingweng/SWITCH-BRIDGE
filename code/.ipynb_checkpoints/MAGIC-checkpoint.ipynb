```python
import magic
import scprep

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import scanpy as sc
```

```python
adata = sc.read_h5ad("discrete_cancer_stages_annotated.h5ad")
adata
```

```python
adata.layers['scvi_normalized']
```

```python
data = pd.DataFrame(adata.layers['scvi_normalized'], index=adata.obs_names, columns=adata.var_names)
data
```

```python
# already normalized by scVI, so library size is constant
scprep.plot.plot_library_size(data, cutoff=1500)
```

```python
# library size set to 1e4
data = scprep.normalize.library_size_normalize(data)
```

```python
# already normalized by scVI, so library size is constant
scprep.plot.plot_library_size(data, cutoff=1500)
```

```python
# log transform the suit scanpy's log-transform
data = scprep.transform.log(data, base='e')
data.head()
```

```python
magic_op = magic.MAGIC(knn=30)
```

```python
data_magic = magic_op.fit_transform(data, genes='all_genes')
```

```python
magic = magic_op.transform(genes="all_genes")
magic.head()
```

```python
magic.to_csv("NE_magic.csv")
```

```python
adata.layers['magic_log'] = magic.to_numpy()
adata
```

```python
# square root transformed and smoothed with MAGIC
adata.write("discrete_cancer_stages_annotated_magic.h5ad")
```
