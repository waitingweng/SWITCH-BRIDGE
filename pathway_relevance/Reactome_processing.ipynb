```python
import pandas as pd
import ast
import sys
sys.path.append('/hdd/edward/the-project/FUNCTIONS')
from methods import module_to_json
```

```python
# Read GMT file (tab-delimited)
gmt_file = "ReactomePathways.gmt"
pathways = []
with open(gmt_file, 'r') as file:
    for line in file:
        data = line.strip().split('\t')
        function = data[0]
        pathway = data[1]
        genes = data[2:]
        if pathway.split("-")[1] == "HSA":
            pathways.append({"function": function, "pathway": pathway, "genes": genes})

# Convert to DataFrame for easier manipulation
pathways_df = pd.DataFrame(pathways)

# Save Homo sapiens pathways
pathways_df.to_csv("ReactomePathways.csv", index=False)
pathways_df
```

```python
Reactome_dict = dict(zip(pathways_df['pathway'], pathways_df['genes']))
Reactome_dict
```

```python
module_to_json(module_dict=Reactome_dict, file_path="ReactomePathways.json")
```
