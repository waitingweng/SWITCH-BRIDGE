use the DEGs that have abs(logFC)>1 as theshold

```python
transitional_set_4S_vs_I = file_to_string(file="transitional_set_4S_vs_I.txt")
transitional_set_4S_vs_III = file_to_string(file="transitional_set_4S_vs_III.txt")
transitional_set_4S_vs_IV = file_to_string(file="transitional_set_4S_vs_IV.txt")
transitional_set_I_vs_III = file_to_string(file="transitional_set_I_vs_III.txt")
transitional_set_I_vs_IV = file_to_string(file="transitional_set_I_vs_IV.txt")
transitional_set_III_vs_IV = file_to_string(file="transitional_set_III_vs_IV.txt")
```

```python
DEG_I_4S = file_to_string(file="../../../results/Cell/deg_I_4S.txt")
DEG_III_4S = file_to_string(file="../../../results/Cell/deg_III_4S.txt")
DEG_IV_4S = file_to_string(file="../../../results/Cell/deg_IV_4S.txt")
DEG_III_I = file_to_string(file="../../../results/Cell/deg_III_I.txt")
DEG_IV_I = file_to_string(file="../../../results/Cell/deg_IV_I.txt")
DEG_IV_III = file_to_string(file="../../../results/Cell/deg_IV_III.txt")
```

```python
targets_I_4S = find_targets(DEG=DEG_I_4S, transitional_set=transitional_set_4S_vs_I)
targets_III_4S = find_targets(DEG=DEG_III_4S, transitional_set=transitional_set_4S_vs_III)
targets_IV_4S = find_targets(DEG=DEG_IV_4S, transitional_set=transitional_set_4S_vs_IV)
targets_III_I = find_targets(DEG=DEG_III_I, transitional_set=transitional_set_I_vs_III)
targets_IV_I = find_targets(DEG=DEG_IV_I, transitional_set=transitional_set_I_vs_IV)
targets_IV_III = find_targets(DEG=DEG_IV_III, transitional_set=transitional_set_III_vs_IV)
```

```python
target_genes = dict(
    targets_I_4S=targets_I_4S,
    targets_III_4S=targets_III_4S,
    targets_IV_4S=targets_IV_4S,
    targets_III_I=targets_III_I,
    targets_IV_I=targets_IV_I,
    targets_IV_III=targets_IV_III
    )
```

```python
module_to_json(module_dict=target_genes, file_path="target_genes.json")
```

```python
with open("target_genes.txt", 'w') as f:
    for i in target_genes:
        for j in target_genes[i]:
            f.write(j + '\n')
```
