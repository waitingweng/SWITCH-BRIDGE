```python
degree_vector = np.array([edge_4S_vs_I, edge_4S_vs_III, edge_4S_vs_IV, edge_I_vs_III, 
                        edge_I_vs_IV, edge_III_vs_IV])
degree_vector
```

```python
from igraph import Graph, plot

# List of directed edges with weights
edges = [("4S", "I", edge_4S_vs_I), ("4S", "III", edge_4S_vs_III), ("IV", "4S", -edge_4S_vs_IV), \
         ("I", "III", edge_I_vs_III), ("I", "IV", edge_I_vs_IV), ("IV", "III", -edge_III_vs_IV)]

# Create a directed weighted graph
graph = Graph.TupleList(edges, directed=True, weights=True, vertex_name_attr="name")

# Print the summary of the graph
print(graph.summary())
```

```python
import matplotlib.pyplot as plt
from igraph import Graph, plot

# Create or load your graph (for demonstration, I'll create a simple one)
# graph = Graph.Ring(10)  # Example graph with 10 vertices connected in a ring

# Layout for the vertices (circle layout)
layout = graph.layout_circle()

# Set vertex colors to red
graph.vs["color"] = "red"

# Format the edge labels to 2 decimal places (assuming weights are already set)
edge_label = graph.es["weight"]
edge_weight = [f"{i:.2f}" for i in edge_label]

# Create a matplotlib figure
fig, ax = plt.subplots()

# Plot the graph with red vertices
plot(
    graph,
    layout=layout,
    vertex_label=graph.vs["name"],
    vertex_label_size=25,
    vertex_size=0.5,  # Adjust as needed
    edge_label=edge_weight,
    edge_label_size=15,
    target=ax  # Integrate with matplotlib axis
)

plt.title("Hotspot", fontsize=20)
plt.savefig("./figures/MDI.svg", bbox_inches="tight")
plt.show()
```

```python
# Save the graph with attributes to a GraphML file
graph.write_graphml("MDI.graphml")
```

```python
conservation_vector
```

```python
np.linalg.norm(conservation_vector)
```

```python
# Calculate cosine similarity
cosine_similarity = np.dot(size_vector, degree_vector) / (np.linalg.norm(size_vector) * np.linalg.norm(degree_vector))
print("Cosine Similarity:", cosine_similarity)
cosine_similarity = np.dot(size_vector, conservation_vector) / (np.linalg.norm(size_vector) * np.linalg.norm(conservation_vector))
print("Cosine Similarity:", cosine_similarity)
cosine_similarity = np.dot(degree_vector, conservation_vector) / (np.linalg.norm(degree_vector) * np.linalg.norm(conservation_vector))
print("Cosine Similarity:", cosine_similarity)
```
