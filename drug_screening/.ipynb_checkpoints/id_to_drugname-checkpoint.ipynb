```python
import xmltodict
import pandas as pd
import numpy as np
```

```python
file = open("full_database.xml")
xml_string = file.read()
file.close()
```

```python
python_dict = xmltodict.parse(xml_string)
```

```python
# python_dict['drugbank'].keys()
```

```python
# np.shape(python_dict['drugbank']['drug'])
```

```python
data = pd.DataFrame.from_dict(python_dict['drugbank']['drug'])
data
```

```python
# data['drugbank-id'][0]
```

```python
# data['drugbank-id'][15560]
```

```python
id = pd.DataFrame(columns=['id'], index=data.index)
id
```

```python
for i in range(16575):
    if isinstance(data['drugbank-id'][i], list):
        id.loc[i] = data['drugbank-id'][i][0]['#text']
    else:
        id.loc[i] = data['drugbank-id'][i]['#text']
```

```python
### bind to the original dataframe
data = pd.concat([data, id], axis = 1)
data
```

```python
!pwd
%cd ../Reciprocal_nlog/
# %cd Cell/Nlog_nabslog/
!pwd
```

```python
I_4S_db = dict()
with open('I_to_4S_-1.96_drugs.txt') as f:
    for line in f.readlines():
        line = line.strip()
        id = line.split(',')[0]
        z_score = float(line.split(',')[1])
        I_4S_db[id] = z_score

I_to_4S = data[data['id'].isin(I_4S_db)].copy()
# Add the proximity_z column based on the mapping
I_to_4S['proximity_z'] = I_to_4S['id'].map(I_4S_db)
I_to_4S.sort_values(by='proximity_z', inplace=True)
I_to_4S.reset_index(drop=True, inplace=True)
I_to_4S.to_csv("I_to_4S.csv", na_rep='NULL')
I_to_4S
```
