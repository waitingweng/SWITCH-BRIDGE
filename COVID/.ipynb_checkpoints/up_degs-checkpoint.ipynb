```python
import pandas as pd
import numpy as np
```

```python
deg_mild_ctrl = pd.read_csv("deg_mild_ctrl.csv", index_col=0)
deg_severe_ctrl = pd.read_csv("deg_severe_ctrl.csv", index_col=0)
deg_severe_mild = pd.read_csv("deg_severe_mild.csv", index_col=0)
```

```python
up_deg_mild_ctrl = deg_mild_ctrl[(deg_mild_ctrl['logfoldchanges'] < -1) & (deg_mild_ctrl['pvals_adj'] < 0.05)]['names']
up_deg_mild_ctrl.to_csv('up_deg_ctrl_mild.txt', header=False, index=False)
up_deg_mild_ctrl
```

```python
up_deg_severe_ctrl = deg_severe_ctrl[(deg_severe_ctrl['logfoldchanges'] < -1) & (deg_severe_ctrl['pvals_adj'] < 0.05)]['names']
up_deg_severe_ctrl.to_csv('up_deg_ctrl_severe.txt', header=False, index=False)
up_deg_severe_ctrl
```

```python
up_deg_severe_mild = deg_severe_mild[(deg_severe_mild['logfoldchanges'] < -1) & (deg_severe_mild['pvals_adj'] < 0.05)]['names']
up_deg_severe_mild.to_csv('up_deg_mild_severe.txt', header=False, index=False)
up_deg_severe_mild
```
