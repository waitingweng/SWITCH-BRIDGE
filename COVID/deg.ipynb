```python
import scanpy as sc
import pandas as pd
import numpy as np
```

```python
adata = sc.read_h5ad("../COVID_all_scVI.h5ad")
adata = adata[adata.obs["cell_type"]=="MoAM"]
```

```python
adata.uns['log1p']["base"] = None
```

```python
# Obtain cluster-specific differentially expressed genes
sc.tl.rank_genes_groups(adata, groupby="CoVID-19 severity", layer='scvi_log', groups=['mild/moderate'], reference='control', method="wilcoxon")
```

```python
deg_mild_ctrl = sc.get.rank_genes_groups_df(adata, group='mild/moderate')
deg_mild_ctrl
```

```python
deg_mild_ctrl.to_csv("deg_mild_ctrl.csv")
```

```python
# Obtain cluster-specific differentially expressed genes
sc.tl.rank_genes_groups(adata, groupby="CoVID-19 severity", layer='scvi_log', groups=['severe/critical'], reference='control', method="wilcoxon")
deg_severe_ctrl = sc.get.rank_genes_groups_df(adata, group='severe/critical')
deg_severe_ctrl.to_csv("deg_severe_ctrl.csv")
deg_severe_ctrl
```

```python
# Obtain cluster-specific differentially expressed genes
sc.tl.rank_genes_groups(adata, groupby="CoVID-19 severity", layer='scvi_log', groups=['severe/critical'], reference='mild/moderate', method="wilcoxon")
deg_severe_mild = sc.get.rank_genes_groups_df(adata, group='severe/critical')
deg_severe_mild.to_csv("deg_severe_mild.csv")
deg_severe_mild
```
