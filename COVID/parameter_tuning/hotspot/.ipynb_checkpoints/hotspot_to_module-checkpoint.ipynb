4S

```python
# Load the Hotspot object from the file
with open("hotspot_object_4S.pkl", "rb") as f:
    hs = pickle.load(f)
```

```python
# fdr_threshold = [0.5, 0.4, 0.3, 0.2, 0.1]
min_gene_threshold = [25,30,35,50,60,80,100,120,160]
for i in min_gene_threshold:
    modules = hs.create_modules(
        min_gene_threshold=i, core_only=True, fdr_threshold=0.05
    )
    module_4S = hotspot_to_module(stage="4S", modules=modules)
    module_to_json(module_dict=module_4S, file_path=f"module_4S_{i}.json")
    # print(modules.value_counts())
```
