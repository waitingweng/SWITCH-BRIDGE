```python
module_size = list()
for i in module_I:
    module_size.append(len(module_I[i]))

import matplotlib.pyplot as plt

# Plot the degree distribution
plt.hist(module_size, bins=range(max(module_size) + 2), density=True, alpha=0.7, color='b', edgecolor='black')
plt.xlabel('Module Size')
plt.ylabel('Frequency')
plt.title('Module Size Distribution')
plt.show()
```

```python
module_size = list()
for i in module_IV:
    module_size.append(len(module_IV[i]))

import matplotlib.pyplot as plt

# Plot the degree distribution
plt.hist(module_size, bins=range(max(module_size) + 2), density=True, alpha=0.7, color='b', edgecolor='black')
plt.xlabel('Module Size')
plt.ylabel('Frequency')
plt.title('Module Size Distribution')
plt.show()
```

```python
intersection_list = list()
for i in module_I:
    for j in module_IV:
        intersection = set(module_I[i]) & set(module_IV[j])
        union = set(module_I[i]) | set(module_IV[j])
        intersection_list.append((i, j, len(intersection)/len(union)))
```

```python
intersection_list[:10]
```

```python
len(intersection_list)
```

```python
import matplotlib.pyplot as plt
import numpy as np
intersection_ratio = [i[2] for i in intersection_list]
# Plot the inintersection_ratio with a bin width of 0.01
plt.hist(intersection_ratio, bins=np.arange(0, max(intersection_ratio) + 0.01, 0.01), density=True, alpha=0.7, color='b', edgecolor='black')
plt.xlabel('Intersection Ratio')
plt.ylabel('Frequency')
plt.title('Intersection Ratio Distribution')
plt.show()
```

```python
# Calculate percentiles
percentiles = np.percentile(intersection_ratio, np.arange(0, 101, 10))

# Display the percentiles
for p, value in zip(range(0, 101, 10), percentiles):
    print(f"{p}th percentile: {value}")

# If you want to visualize the percentiles, you can use a boxplot
import matplotlib.pyplot as plt

plt.boxplot(intersection_ratio, vert=False)
plt.xticks(percentiles)
plt.xlabel('Intersection Ratio')
plt.ylabel('Percentiles')
plt.title('Boxplot of Intersection Ratio Distribution with Percentiles')
plt.show()
```

```python
# Iterate in reverse order to prevent index out of range
for i in range(len(intersection_list)-1, -1, -1):
    if intersection_list[i][2] == 0:
        intersection_list.pop(i)
```
