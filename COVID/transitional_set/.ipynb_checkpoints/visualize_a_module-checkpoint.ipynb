```python
def module_to_dict(file=""):
    # importing the module
    import json
     
    # Opening JSON file
    with open(file, 'r') as json_file:
        data = json.load(json_file)
        return data  
```

```python
parent = set(module_I["module_I_27"])
# hidden_children1 = set(module_I["module_I_9"])
# hidden_children2 = set(module_IV["module_IV_82"])
child1 = set(module_III["module_III_5"])
overlap_child1 = parent & child1
child2 = set(module_III["module_III_24"])
overlap_child2 = parent & child2
print(len(parent), len(child1), len(child2), len(overlap_child1), len(overlap_child2))
```

```python
pcc_I = pd.read_csv("/home/jovyan/work/results/Cell/pcc_I.csv", index_col=0)
pcc_III = pd.read_csv("/home/jovyan/work/results/Cell/pcc_III.csv", index_col=0)
```

```python
parent_pcc = pcc_I.loc[list(parent), list(parent)]
child1_pcc = pcc_III.loc[list(child1), list(child1)]
child2_pcc = pcc_III.loc[list(child2), list(child2)]
print(parent_pcc.shape, child1_pcc.shape, child2_pcc.shape)
```

draw parent module

```python
# Threshold for including edges
threshold = 0.3

gene_symbols = parent_pcc.index
# Convert PCC matrix to binary adjacency matrix
adjacency_matrix = (abs(parent_pcc) > threshold).astype(int)
# Create an unweighted graph
parent_graph = Graph.Adjacency(adjacency_matrix, mode="UNDIRECTED", loops=False)

# Set vertex names
parent_graph.vs["name"] = gene_symbols
for i in parent_graph.vs:
    if i["name"] in transitional_set_I_vs_III:
        i["transitional_set"] = i["name"]
    else:
        i["transitional_set"] = ""
        
colors = list()
for i in gene_symbols:
    if i in overlap_child1:
        colors.append("blue")
    elif i in overlap_child2:
        colors.append("red")
    # elif i in hidden_children1:
    #     colors.append("yellow")
    # elif i in hidden_children2:
    #     colors.append("brown")
    else:
        colors.append("black")
parent_graph.vs["color"] = colors

# Annotate edges with a "color" attribute based on node colors
edge_colors = list()
for edge in parent_graph.es:
    source_color = parent_graph.vs[edge.source]["color"]
    target_color = parent_graph.vs[edge.target]["color"]
    
    # Set edge color based on node colors
    if source_color == "black" and target_color == "black":
        edge_colors.append("#CCCCCC")
    else:
        edge_colors.append("green")

# Add the "color" attribute to the edges
parent_graph.es["color"] = edge_colors


# Display the graph summary
print(parent_graph.summary())
parent_graph.is_connected()
```

```python
gv.vis(parent_graph, show_node_label=True, show_edge_label=False, node_label_data_source="name")
```

```python
gv.vis(parent_graph, show_node_label=True, show_edge_label=False, node_label_data_source="transitional_set")
```

draw children module

```python
# Threshold for including edges
threshold = 0.3

gene_symbols = child1_pcc.index

# Convert PCC matrix to binary adjacency matrix
adjacency_matrix = (abs(child1_pcc) > threshold).astype(int)

# Create an unweighted graph
child1_graph = Graph.Adjacency(adjacency_matrix, mode="UNDIRECTED", loops=False)

# Set vertex names
child1_graph.vs["name"] = gene_symbols
for i in child1_graph.vs:
    if i["name"] in transitional_set_I_vs_III:
        i["transitional_set"] = i["name"]
    else:
        i["transitional_set"] = ""

colors = list()
for i in gene_symbols:
    if i in overlap_child1:
        colors.append("blue")
    else:
        colors.append("orange")
child1_graph.vs["color"] = colors

# Display the graph summary
print(child1_graph.summary())
child1_graph.is_connected()
```

```python
gv.vis(child1_graph, show_node_label=True, show_edge_label=False, node_label_data_source='name')
```

```python
# Threshold for including edges
threshold = 0.3

gene_symbols = child2_pcc.index

# Convert PCC matrix to binary adjacency matrix
adjacency_matrix = (abs(child2_pcc) > threshold).astype(int)

# Create an unweighted graph
child2_graph = Graph.Adjacency(adjacency_matrix, mode="UNDIRECTED", loops=False)

# Set vertex names
child2_graph.vs["name"] = gene_symbols

colors = list()
for i in gene_symbols:
    if i in overlap_child2:
        colors.append("red")
    else:
        colors.append("green")
child2_graph.vs["color"] = colors

# Display the graph summary
print(child2_graph.summary())
child2_graph.is_connected()
```
