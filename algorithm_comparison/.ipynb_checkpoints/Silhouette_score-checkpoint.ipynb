Silhouette score evaluation

```python
import pandas as pd
import numpy as np
```

```python
# import the PCC data
pcc_4S = pd.read_csv("/home/jovyan/work/PCC_Module_method/results/pcc_4S.csv", index_col=0)
pcc_I = pd.read_csv("/home/jovyan/work/PCC_Module_method/results/pcc_I.csv", index_col=0)
pcc_III = pd.read_csv("/home/jovyan/work/PCC_Module_method/results/pcc_III.csv", index_col=0)
pcc_IV = pd.read_csv("/home/jovyan/work/PCC_Module_method/results/pcc_IV.csv", index_col=0)
pcc_NA = pd.read_csv("/home/jovyan/work/PCC_Module_method/results/pcc_NA.csv", index_col=0)
pcc_4S
```

```python
def module_to_dict(file=""):
    # importing the module
    import json
     
    # Opening JSON file
    with open(file, 'r') as json_file:
        data = json.load(json_file)
        return data  
```

```python
def correlation_distance(pcc_matrix):
    # Compute the correlation distance matrix
    correlation_distance_matrix = 1 - np.abs(pcc_matrix)
    
    # Ensure non-negative values
    correlation_distance_matrix[correlation_distance_matrix < 0] = 0

    return correlation_distance_matrix
```

```python
distance_matrix_4S = correlation_distance(pcc_matrix=pcc_4S)
distance_matrix_I = correlation_distance(pcc_matrix=pcc_I)
distance_matrix_III = correlation_distance(pcc_matrix=pcc_III)
distance_matrix_IV = correlation_distance(pcc_matrix=pcc_IV)
distance_matrix_NA = correlation_distance(pcc_matrix=pcc_NA)
```

```python
def silhouette_score(distance_matrix, modules):
    module_genes = list()
    for i in modules.values():
        if len(i) >= 10:
            for j in i:
                module_genes.append(j)
    module_distance_matrix = distance_matrix.loc[module_genes, module_genes]
    print(module_distance_matrix.shape)

    cluster_labels = list()
    for i in module_distance_matrix.index:
        for name in modules:
            if i in modules[name]:
                cluster_labels.append(name)
                break
    if len(set(cluster_labels)) == 1:
        print("Silhouette Score: undefined due to only 1 cluster found")
        return
    from sklearn.metrics import silhouette_score
    # Calculate Silhouette score
    silhouette_avg = silhouette_score(module_distance_matrix, cluster_labels)
    
    print("Silhouette Score:", silhouette_avg)
    return silhouette_avg
```

```python
def silhouette_score_to_txt(path, silhouette_score_list):
    with open(path, 'w') as file:
        for i in silhouette_score_list:
            file.write(str(i) + "\n")
```

METHOD2

```python
module_4S = module_to_dict("/home/jovyan/work/METHOD2_leiden/module_4S.json")
module_I = module_to_dict("/home/jovyan/work/METHOD2_leiden/module_I.json")
module_III = module_to_dict("/home/jovyan/work/METHOD2_leiden/module_III.json")
module_IV = module_to_dict("/home/jovyan/work/METHOD2_leiden/module_IV.json")
module_NA = module_to_dict("/home/jovyan/work/METHOD2_leiden/module_NA.json")
silhouette_score_4S = silhouette_score(distance_matrix=distance_matrix_4S, modules=module_4S)
silhouette_score_I = silhouette_score(distance_matrix=distance_matrix_I, modules=module_I)
silhouette_score_III = silhouette_score(distance_matrix=distance_matrix_III, modules=module_III)
silhouette_score_IV = silhouette_score(distance_matrix=distance_matrix_IV, modules=module_IV)
silhouette_score_NA = silhouette_score(distance_matrix=distance_matrix_NA, modules=module_NA)
silhouette_score_to_txt(path="method2_silhouette_score.txt", silhouette_score_list=[silhouette_score_4S, silhouette_score_I, silhouette_score_III, silhouette_score_IV, silhouette_score_NA])
```
