```python
# get the number of modules at the resolution
resolution0_1 = list()
resolution1_2 = list()
resolution2_3 = list()
resolution3_4 = list()
resolution4_5 = list()
resolution5_6 = list()
resolution6_7 = list()
resolution7_8 = list()
```

```python
def subtraction(list1, list2):
    return [a - b for a, b in zip(list1, list2)]
```

```python
def add_module_numbers(input_path, reverse_order = False):
    files = glob.glob(input_path + "*.json")
    print(files)
    
    resolution_dict = dict()
    for f in files:
        number_of_modules = len(module_to_dict(f))
        resolution = float(f.split("_")[-1][:-5])
        if reverse_order == True:
            resolution = -resolution
        if resolution in resolution_dict:
            resolution_dict[resolution].append(number_of_modules)
        else:
            resolution_dict[resolution] = [number_of_modules]
    
    sorted_resolution = list(dict(sorted(resolution_dict.items())).values())
    
    # print(sorted_resolution)
    global resolution0_1, resolution1_2, resolution2_3, resolution3_4, resolution4_5, resolution5_6, resolution6_7, resolution7_8
    for k in range(len(sorted_resolution)-1):
        print(sorted_resolution[k])
        if k == 0:
            resolution0_1 += subtraction(sorted_resolution[k+1], sorted_resolution[k])
        elif k == 1:
            resolution1_2 += subtraction(sorted_resolution[k+1], sorted_resolution[k])
        elif k == 2:
            resolution2_3 += subtraction(sorted_resolution[k+1], sorted_resolution[k])
        elif k == 3:
            resolution3_4 += subtraction(sorted_resolution[k+1], sorted_resolution[k])
        elif k == 4:
            resolution4_5 += subtraction(sorted_resolution[k+1], sorted_resolution[k])
        elif k == 5:
            resolution5_6 += subtraction(sorted_resolution[k+1], sorted_resolution[k])
        elif k == 6:
            resolution6_7 += subtraction(sorted_resolution[k+1], sorted_resolution[k])
        elif k == 7:
            resolution7_8 += subtraction(sorted_resolution[k+1], sorted_resolution[k])
    print()
```

```python
SWIMs = ["PCC_SWIM", "Binary_SWIM", "Linear_SWIM", "Nlog_SWIM", "Reciprocal_SWIM"]
knds = ["knd_binary", "knd_linear", "knd_nlog", "knd_reciprocal", "knn"]
```

```python
for i in range(len(SWIMs)):
    for j in range(len(knds)):
        input_path = f"{SWIMs[i]}/{knds[j]}/Cell/"
        add_module_numbers(input_path)
```

```python
controls = ["PCC_Leiden", "hdWGCNA", "Hotspot"]
```

```python
for i in range(len(controls)):
    if controls[i] == "PCC_Leiden":
        input_path = f"{controls[i]}/Cell/"
        add_module_numbers(input_path)
    elif controls[i] == "hdWGCNA":
        input_path = f"{controls[i]}/Cell/hdWGCNA/"
        add_module_numbers(input_path)
    elif controls[i] == "Hotspot":
        input_path = f"{controls[i]}/Cell/"
        add_module_numbers(input_path, reverse_order=True)
```
