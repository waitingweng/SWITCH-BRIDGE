```python
resolution_values = [1,2,3,4,5,10,15,20,25]  # for n_neighbors=30: 15~320 clusters
# n_Neighbors doesn't affect the number of clusters at all, so don't need to use a large value
# resolution_values = [5,10,15,20,25,30,35,40,45]  # for n_neighbors=50: 60~530 clusters
# resolution_values = [5,10,15,20,25,30,35,40,45]  # for n_neighbors=100: 50~540 clusters

for resolution_parameter in resolution_values:
    pcc_to_leiden(data=pcc_4S, data_name="pcc_4S", n_neighbors=30, resolution=resolution_parameter, fig_path=f"./figures/degree_distribution_4S_{resolution_parameter}.svg")
    module_4S = leiden_to_module(pcc_4S, min_module_size=4)
    module_to_json(module_dict=module_4S, file_path=f"module_4S_{resolution_parameter}.json")
```
