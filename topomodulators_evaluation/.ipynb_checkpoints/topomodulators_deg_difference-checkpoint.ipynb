```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Step 1: Example sets of genes
set_names = ["C_s", "deg", "N_n", "N_r", "R_n", "R_r", "P_l", "hd", "Hot"]
sets = [curated_set, deg, Nlog_nlog_target_set, Nlog_reciprocal_target_set,
       Reciprocal_nlog_target_set, Reciprocal_reciprocal_target_set, PCC_Leiden_target_set,
       hdWGCNA_target_set, Hotspot_target_set]

# Step 2–3: Compute intersection matrix
intersection_matrix = np.zeros((9, 9), dtype=int)
for i in range(9):
    for j in range(9):
        intersection_matrix[i][j] = len(sets[i] & sets[j])

# Step 4: Plot heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(intersection_matrix, annot=True, fmt="d", cmap="coolwarm", xticklabels=set_names, yticklabels=set_names)
plt.title("Intersection of Target Genes", fontsize=20)
plt.savefig("./figures/target_genes_intersection.svg", bbox_inches="tight")
plt.show()
```
