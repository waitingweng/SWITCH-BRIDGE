Bell_97

```python
with open("background_Bell_97_R2.txt", 'r') as f:
    text = f.read().split('\n')
    ratio = text[8].split(' ')[-1]
    sig_survival_set, background_set = map(int, ratio.split('/'))
    print(sig_survival_set, background_set)
    print(sig_survival_set/background_set)
```

test how the curated set performs (compare to SWITCH)

```python
with open("curated_Bell_97_R2.txt", 'r') as f:
    text = f.read().split('\n')
    ratio = text[8].split(' ')[-1]
    sig_survival_target_set, target_set = map(int, ratio.split('/'))
    print(sig_survival_target_set, target_set)
    print(sig_survival_target_set/target_set)
```

```python
# Define your variables
N = background_set  # Total genes in the universe
K = sig_survival_set   # Genes in the curated list
n = target_set     # Genes in your prediction set
k = sig_survival_target_set   # Overlapping genes

# Calculate the p-value for enrichment
p_value = hypergeom.sf(k-1, N, K, n)  # sf = 1 - cdf
print("P-value:", p_value)
```
